```python
# Create a BankAccount class with:
# deposit()
# withdraw()
# check_balance()

class BankAccount:
    T_amount = 0
    def __init__(self):
        pass
    def desposit(self , D_amount):
        print(f"Desposited Amount : {D_amount}")
        self.T_amount += D_amount

    def withdraw(self , W_amount) :
        print(f"Withdraw Amount : {W_amount}")
        self.T_amount -= W_amount

    def A_balance(self):
        print(f"Total Balance is : {self.T_amount}")

b1 = BankAccount()

b1.desposit(50000)
b1.desposit(34000)
b1.withdraw(23000)

b1.A_balance()    
```

```python
# Create a Person class and inherit it into a Student class.
class Person:
    def __init__(self , name , age):
        self.name = name
        self.age  = age 

class Student(Person):
    def __init__(self , name , age , Deg , rollNo):
        super().__init__(name , age)
        self.Deg = Deg
        self.rollNo = rollNo

    def Info(self):
        print(f"Name : {self.name}")
        print(f"Age : {self.age}")
        print(f"Degree : {self.Deg}")
        print(f"RollNo : {self.rollNo}")

S1 = Student("Ali" , 25 , "Bscs" , "2020f-mulbscs-109")
S1.Info()
```

```python
# Create an Animal class and override a sound() method
class Animal :
    def __init__(self , name , color):
        self.name = name
        self.color = color

    def sound():
        print("Animal have a Sound")

class Dog(Animal):
    def __init__(self , name , color , price):
        super().__init__(name , color)
        self.price = price
    def info(self):
        print(f"Name : {self.name}")
        print(f"Color : {self.color}")
        print(f"Price : {self.price}")
    def sound(self):
        print("Dog can Bark")

class Parrot(Animal):

    def __init__(self,name , color , price ):
        super().__init__(name , color)
        self.price = price

    def info(self):
        print(f"Name : {self.name}")
        print(f"Color : {self.color}")
        print(f"Price : {self.price}")
    def sound(self):
        print("Parrot have sound like Person")
        
d1 = Dog("Boddy" , "Black" , 23000)
d1.sound()

p1 = Parrot("Moon" , "Green" , 20000)
p1.sound()
```

```python
class Employees :
    def __init__(self):
        self.emp = []

    def addEmp(self , name):
        print(f"Employee : {name} is Added")
        self.emp.append(name)

    def remove(self , name):
        if name not in self.emp:
            print("Employee does not exist")
        self.emp.remove(name)

    def show(abs):
        if len(abs.emp)==0:
            print("No Employeee")
        print("\n Employee List is ....")
        for i in abs.emp:
            print(i)

    def seEmp(abs,empname ):
        if empname in abs.emp:
            print(f"Employee is : {empname}")
        else :
            print(f"Employee not Found")

e1 = Employees()
e1.addEmp("Wafadar")
e1.addEmp("Ghayur")
e1.addEmp("Qasim")
e1.show()      
e1.addEmp("Zaman")
e1.addEmp("Nadeem") 
e1.addEmp("Johar") 
e1.seEmp("Nadeem")
e1.remove("Johar")
e1.show()
```
