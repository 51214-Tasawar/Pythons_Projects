```python
class Triangle:
    def __init__(self , Length , Width):
        self.Length = Length
        self.Width  =  Width
    def Area(this):
        print(f"Area is : {1/2 * (this.Length * this.Width)}")

class Reactangle :
    def __init__(self , Length , Width):
        self.Length = Length
        self.Width = Width
    def Area(this):
        print(f"Area is {this.Length * this.Width}")
T1 = Triangle(12,14)
R1 = Reactangle(12,14)
T1.Area()
R1.Area()
```

```python
class Vehicle :
    def __init__(self ,  color ):
        self.color = color

    def Move(this):
        print("Driving Car")

class Car(Vehicle):
    pass

class Plane(Vehicle):
    def Move(this):
        print("Flying A Plane")

class Train(Vehicle):
    def Move(this):
        print("Train on Track")

C1 = Car("White")
P1 = Plane("Light Gray")
T1 = Train("Green")
for i in (C1 , P1 , T1) :
    print(f"Color is : {i.color}")
    i.Move()
    print("... Next...")
```
