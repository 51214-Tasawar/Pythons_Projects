```python
class empdata :
    user = {
        "name" : "" ,
        "age " : 0
    }
    def __init__(self):
        self.employees = []
    def add_Emp(self, user) :
        self.employees.append(user)
        print(f"Added User :{user}")

    def remov_Emp(self , user):
        if user in self.employees:
            self.employees.remove(user)
            print(f"{user} removed")

    def show_User(self ) :
        print("All Users are ")
        for i in self.employees :
            print(i)
users = empdata()
users.add_Emp({"name" : "Ali" , "age" : 22})
users.add_Emp({"name": "Wafadar" , "age" : 26})
users.add_Emp({"name": "Ghayur" , "age" : 25})
users.add_Emp({"name": "Farhat" , "age" : 23})
users.add_Emp({"name": "Zarak" , "age" : 23})
users.add_Emp({"name": "Habil" , "age" : 22})
users.add_Emp({"name": "Tasawar" , "age" : 25})
users.show_User()
users.remov_Emp({"name": "Tasawar" , "age" : 25})
users.show_User()
```
