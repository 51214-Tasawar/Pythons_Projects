```python
class info:
    name = "Ali Khan" 
    age  = 23
    Add = "Multan"
s1 = info()
print("Name : " , s1.name)
print("Age : " , s1.age)
print("Address : " , s1.Add)
```

```python
class User :
    name = "Hassan"
    age = 12 
    Add = "Multan"
s1 = User()
s1.name = "Wafadar"
print(s1.name)
print(s1.age)
print(s1.Add)
```

```python
s2 = User()
s2.name = "Ghayur"
s2.age = 25
print(f"Name : {s2.name} Age : {s2.age}  Address : {s2.Add}")
```
