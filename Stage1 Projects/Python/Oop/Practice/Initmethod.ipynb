```python
class student :
    def __init__(self , name , age):
        self.name = name
        self.age = age
    def display(self):
        print(f"Name : {self.name} Age : {self.age}")
s1 = student("Hassan" , 20)
s1.display()
```

```python
class emp :
    def __init__(self , name , depart , desination , salary , age = 20):
        self.name = name
        self.depart = depart
        self.desination = desination
        self.salary = salary
        self.age = age
    def Display(self):
        print(f"Name : {self.name} Department : {self.depart}")
        print(f"Desination :{self.desination} Salary : {self.salary}")
        print(f"Age : {self.age}")
e1 = emp("Ghayur" , "Dev Team" , "Junior AI Developer" , 50000 , 30)
e1.Display()
e2 = emp("Farhat" , "Dev Team" , "Internee" , 250000)
e2.Display()
```

```python
class Pet :
    def __init__(self , name , color):
        self.name = name
        self.color = color
    def Res(ABS):
        return "Name "+ABS.name+" Color "+ ABS.color
    def display(ab):
        petinfo = ab.Res()
        print("Below is Pet Info ...")
        print(petinfo)
p1 = Pet("Buddy" , "Black")
p1.display()
```

```python
class Math :
    def __init__(self , a , b):
        self.a = a 
        self.b = b
    def add(self):
        return self.a + self.b
    def Sub(self) :
        return self.a - self.b
    def Mul(self) :
        return self.a * self.b
    def Div(self) :
        return self.a / self.b
    def display(self) :
        print(f"Addition : {self.add()}")
        print(f"Substraction : {self.Sub()}")
        print(f"Multiplaction : {self.Mul()}")
        print(f"Division : {self.Div()}")
m1 = Math(12,4)
m1.display()
    
```
