```python
# Find the factorial of a number.
num = int(input("Enter Number"))
fac = 1
for i in range(1 , num+1):
    fac *= i
print(fac)
```

```python
# Reverse a number using a loop.
num = int(input("Enter Number"))
revnum = 0
for i in range(len(str(num))):
    digi = num % 10 
    revnum = revnum * 10 + digi
    num = num  // 10
print(revnum)
```

```python
#find largest digit in number 
num = int(input("Enter Number"))
largenum = 0
for i in range(len(str(num))):
    digi = num % 10 
    if largenum < digi :
        largenum = digi
    num = num  // 10
print(largenum)
```

```python
# Fibonacci  for given term 
Term = int(input("Enter Num"))
pre = 0
next = 1
for i in range(1 , Term+1):
     temp = pre + next
     print(pre)
     pre = next
     next = temp
```

```python
# Check Plindrome number 
num = int(input("Enter Number"))
number = num
revnum = 0
for i in range(len(str(num))):
    digi = num % 10 
    revnum = revnum * 10 + digi
    num = num  // 10
if revnum == number :
    print("Plindrome Number ")
else :
    print("Not Plindrome")
```

```python
# Check number is Armstrong
num = int(input("Enter Number "))
pow = result  = 0
x = y = num
for i in range(len(str(x))):
    pow += 1
    x = x // 10

for i in range(len(str(y))) :
    newpow = pow 
    res = 1
    digi = y % 10
    for j in range(newpow , 0 , -1) :
        res *= digi
    result += res
    y = y // 10 
if result == num :
    print("Prime")
else :
    print("Not A Prime")

```
