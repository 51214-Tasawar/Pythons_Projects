```python
# Reverse a number using a loop.
num = int(input("Enter Number"))
reverseNum = 0
while num != 0 :
    digi = num % 10 
    reverseNum = reverseNum * 10 + digi
    num = num // 10
print(f"Reverse Number : {reverseNum}")
```

```python
#find largest digit in number 
num = int(input("Enter Number"))
Lagerdigi = 0
while num != 0 :
    digi = num % 10 
    if Lagerdigi < digi :
        Lagerdigi = digi
    num = num // 10
print(f"Largest Digit : {Lagerdigi}")
```

```python
# Check Plindrome number 
num = int(input("Enter Number"))
number = num
reverseNum = 0
while num != 0 :
    digi = num % 10 
    reverseNum = reverseNum * 10 + digi
    num = num // 10
if number == reverseNum :
    print(f"Number is Plindrome : {number}")
else :
    print("Not Plindrome")
```

```python
# Check Armstrong Number : ?
num = int(input("Enter number "))
x = y = num
result = 0
pow = 0

while x != 0 :
    pow += 1
    x = x // 10

while y != 0 :
    newpow = pow
    res = 1
    digi = y % 10 
    while newpow != 0 :
        res *= digi
        newpow -= 1
    result += res
    y = y // 10
if num == result :
    print("Number is ArmStrong")
else :
    print("Not Armstrong")
```

```python
# Print:
#     *
#    **
#   ***
#  ****
# *****
n = 5
row = 1
while row <= n:
    space = 1
    while space <= n - row:
        print(" ", end="")
        space += 1
    star = 1
    while star <= row:
        print("*", end="")
        star += 1

    print()
    row += 1
```

```python
# Find whether a number is prime.
num = int(input("Enter Number"))
is_Prime = True
i = 2
while i < num :
    if num % i == 0 :
        is_Prime = False
        break
    i += 1
if is_Prime :
    print("Prime Number")
else :
    print("Not Prime")
 
```
