```python
def users(*agrs) :
    print(agrs)
    print(agrs[2])
users("ali" , "Manzoor" , "Qasim"  , "Jami")
```

```python
def Sum(*num) :
    total = 0
    for i in num :
        total += i 
    print(f"Total is : {total}")

Sum(1,2,3,43,4,5,67)
Sum(12,13,14,12,13,50,90)
Sum(10,20,30,40)
```

```python
#Practice **Kwargs

def info(**user):
    print(user["fname"]+ " " + user["lname"])

info(fname = "Wafadar",email="abc@gmail.com" , lname = "Khan")
```

```python
def DevTeam(**info) :
    print(f"Name : {info["name"]} Salary : {info["salary"]}")
DevTeam(name = "Ghayur" , salary = 90000)
```

```python
def grandFun():
    global x 
    x = 30
    print(x)
    def parentfun():
        x = 40
        print(x)
        def childfun():
            x = 90 
            print(x)
        childfun()
    parentfun()
grandFun()
print(x)
```
