```python
# What is a dot product ?
# The dot product is a mathematical operation that takes two vectors, multiplies their corresponding elements,
#  and adds the results.
# For two vectors:

# A=[a_1,a_2,a_3]and B=[b_1,b_2,b_3]

# The dot product is:

# A.B=a_1b_1+a_2b_2+a_3b_3

# In simple words:
# 1. Multiply the first elements.
# 2. Multiply the second elements.
# 3. Multiply the third elements.
# 4. Add all the products.
# The result is a single number (scalar).

```

```python
# Why is dot product important in machine learning?

# The dot product is very important in Machine Learning because
# it allows a model to combine input features with their weights and produce a result.

# Simple example
# Suppose we have a house with 3 features:
#  X=[size,bedrooms,age] 
# After converting them to numbers:
# X=[1000,3,5] 

# The model has learned weights:
# W=[200,5000,-1000] 
# The dot product is:
# W.X=(1000*200)+(3*5000)+(5*1000) 
# =200000+15000-5000 =210000 
# This gives a value that can be used to predict something like a house price.



# Why ML uses it
# The basic idea is:
# Output= X.W + b

# where:
# X= input features
# W= weights learned by the model
# b= bias
# X.W= dot product

# This appears in linear regression, logistic regression, neural networks, and many other ML algorithms.

# Think of it this way
# The dot product is basically asking:
# "How strongly do these inputs and their weights combine to produce an output?"

# For a neural network, you can have:
# [x1​,x2​,x3​] ⋅ [w1​,w2​,w3​] =x1​w1​+x2​w2​+x3​w3​ 

# Then the result is usually passed through an activation function.

# So, when learning ML, remember:
# Features*Weights→Dot Product→Prediction​
```

```python
# How is dot product used to measure similarity ?
# The dot product can measure similarity by checking how much two vectors point in the same direction.

# Suppose:
# A=[1,2]
# B=[2,4]
# Calculate the dot product:
# A⋅B=(1×2)+(2×4)
# =2+8=10​
# Because B points in the same direction as A their dot product is positive and relatively large.


# Direction matters
# For two vectors:
# A⋅B=∣A∣∣B∣cos(θ)
# where theta is the angle between them.

# Angle	Dot product	Meaning
# (0)	Large positive	Very similar direction
# (90)	0	No directional similarity
# (180)	Negative	Opposite directions

# Dot product alone depends on both direction and vector size.
# Cosine similarity removes the effect of size and focuses mainly on direction.

# In Machine Learning

# Imagine words are represented as vectors:

# Cat=[0.8,0.7,0.1]
# Kitten=[0.7,0.8,0.2]

# Their dot product will be positive and relatively large because their representations point in similar directions.

# This is useful for things like:

# 🔎 Search — matching a query with documents
# 🎬 Recommendation systems — finding similar items
# 🤖 AI embeddings — comparing meanings of text
# 🖼️ Image recognition — comparing image representations

# Important: Raw dot product is affected by vector length. For a pure similarity measure, ML systems often use 
# cosine similarity, which normalizes the vectors:
```

```python
# consine simalrity :
# A.B = |A| |B| cos(θ)
# cos(θ) = A.B/ |A| |B| 

# The result is normally between -1 and +1:

# Cosine similarity	Direction
# 1	Exactly same direction
# 0	90° / perpendicular
# -1	Exactly opposite direction
# 0.8	Very similar direction
# 0.2	Somewhat similar
# -0.5	Opposite-ish direction

# Suppose :
# Sentence A → [0.21, 0.75, 0.13, 0.91]
# Sentence B → [0.22, 0.73, 0.15, 0.89]

# We can calculate their cosine similarity.
# If the result is close to 1, their vectors point in similar directions, so the AI considers their
# embeddings semantically similar.
```

```python
# if A and B are equals ?
# If A and B are equal vectors, 
# for example:
# A=[1,2]
# B=[1,2]
# then they point in exactly the same direction.

# Calculate cosine similarity

# First, dot product:

# A⋅B=(1×1)+(2×2)
# =1+4=5

# Magnitude of A:
# ∣A∣=underroot(1+4)
# ​=underroot(5)
# ​

# Magnitude of B:
# ∣A∣=underroot(1+4)
# ​=underroot(5)


# Now:
# cons(θ) = 5 / underroot(25)
# cons(θ) = 5/5
# cons(θ) = 1


# Important

# If:
# A=B
# then:
# Cosine similarity=1

# because the vectors have exactly the same direction.
# Even if the values aren't equal but one is a positive multiple of the other:
#  A=[1,2] B=[2,4] 

# the cosine similarity is also 1, because they still point in the same direction.
```

```python
# What happens if two vectors point in very different directions ?
# If two vectors point in very different directions, their dot product becomes small
# or negative, depending on the angle.
# for example :

# A=[1,0],B=[0,1]
# Their dot product is:
# A.B=1*0+0*1=0

# A.B = |A| |B| cos(θ)
# cos(θ) = A.B/ |A| |B| 

# Expamles :

# 1. Same direction — 0∘
#   cos(0∘)=1
# Dot product is positive and large → strong similarity.

# 2. Perpendicular — 90∘
#    cos(90∘)=0
# A⋅B=0
# → No directional similarity.

# 3.Opposite directions — 180∘
#    cos(180∘)=−1
# Dot product is negative → vectors point in opposite directions.

# Visual idea

# Imagine arrows:

# →→
# Same direction → high positive dot product.

# →↑
# Very different direction → dot product approaches 0.

# →←
# Opposite direction → negative dot product.


# More different directions⇒smaller dot product

# and when they become exactly opposite:
# dot product is negative
```

```python
# Why is dot product important in embeddings ?
# An embedding is a vector representation of data, such as a sentence, image, or document.
# AI systems can calculate the dot product between embeddings to obtain a similarity-related score.
# For example, suppose:
# Q=[0.2,0.8] , D=[0.3,0.9]
# Here, (Q) is a query embedding and (D) is a document embedding.

# Their dot product is:
# Q.D=(0.2*0.3)+(0.8*0.9)
# =0.06+0.72=0.78
# This score can be used to compare the query and document, depending
# on how the embeddings were generated and normalized.

# What happens if two vectors are highly similar  ?
# If two vectors point in similar directions, their cosine similarity will be close to 1.

# For example:

# A=[1,2],B=[2,4]
# Both vectors point in the same direction, so their cosine similarity is 1.
# In an AI search system, if a query embedding and a document embedding have high similarity, the document
#  may be considered relevant to the query.
# For example:
# - Query: "How can I learn Python?"
# - Document: "A beginner's guide to learning Python."
# Their embeddings may be similar because the sentences have related meanings.
# Note that a high raw dot product does not always mean high similarity unless vector magnitudes are 
# accounted for or the embedding system is designed for dot-product scoring.


```
