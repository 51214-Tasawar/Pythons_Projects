```python
# What does standard deviation measure   ?
# Standard deviation measures how spread out values are from their mean (average).
# Standard deviation measures how spread out the values are from their mean.

# Small standard deviation → values are close to the mean.
# Large standard deviation → values are far from the mean.
# Zero standard deviation → all values are identical.
```

```python
# 2. Calculate the standard deviation of 2, 4, 6  ?
# First find the mean:
# mu=2+4+6/3=4 
# Now find how far each value is from the mean:
# 2-4=-2  4-4=0 6-4=2
# Square the differences:
# (−2)
# 2
# =4
# 0^2=0 
# 2
# 2
# =4
# Add them:
# 4+0+4=8 
# For population standard deviation, divide by 3:
# Variance=8/3	​
# Then take the square root:
# Standard Deviation= sqrt{8/3}
# SDapprox1.63
# So, the standard deviation of 2,4,6is approximately 1.63 using the population formula.
```

```python
# # Difference between varience and bstandrard deviation   ?
# The main difference is that variance and standard deviation both measure spread, but standard deviation is
# easier to understand because it is in the same unit as the original data.

# For example, if your data is height in centimeters:
# Variance → cm²
# Standard deviation → cm
# That's why standard deviation is often easier to interpret.
# Remember
# ariance = SD^2
# SD= sqrt(Variance)
```
