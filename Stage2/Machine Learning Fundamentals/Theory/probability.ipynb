```python
# 1. What is probability  ?
# Probability is a measure of how likely an event is to happen.
# Its value ranges from 0 to 1.
# 0<= p(A) <= 1
# The formula for probability is:
# P(A)=Number of favorable outcomes /Total number of equally likely outcomes
```

```python
# What is the probability of getting heads from a fair coin  ?
# A fair coin has two equally likely outcomes:
# - Heads
# - Tails
# Therefore,
# P(Heads)=1/2=0.5
# The probability is 50%.
```

```python
# What is the probability of rolling a 6 on a normal die?
# A normal fair die has six possible outcomes: 1, 2, 3, 4, 5, and 6.
# Only one outcome gives us a 6.
# P(6)=1/6=0.167

# This is approximately 16.67%.
```

```python
# What is the difference between probability 0 and probability 1  ?

# Probability 0 The event is impossible under the assumed conditions.
# Example: Rolling a 7 on a standard six-sided die.

# Probability 1 The event is certain under the assumed conditions.  
# Example: Rolling a number from 1 to 6 on a standard six-sided die.
```

```python
# What does probability 0.8 mean  ?
# A probability of 0.8 means an 80% chance of an event occurring.
# 0.8 * 100 = 80%
# For example, if a fair process has a probability of 0.8 for an event, that event would be expected to 
# occur about 80 times in 100 independent trials, on average over the long run.
# It does not guarantee that the event will happen exactly 80 times.
```

```python
# if an AI model gives these probabilities, what does this mean?
# cat = 0.85
# dog = 0.10
# car = 0.05


# These values represent the model's predicted probabilities for three possible classes.
# Class	   Probability      	Percentage
# Cat      	 0.85                	85%
# Dog	     0.10	                10%
# Car	     0.05	                 5%
# Total      1.00               	100%
# Explanation
# - Cat (0.85): The model assigns an 85% probability to the image being a cat.  
# - Dog (0.10): The model assigns a 10% probability to it being a dog.
# - Car (0.05): The model assigns a 5% probability to it being a car.
# The model's most likely prediction is cat, because it has the highest probability.
```

```python
# Why do classification models often output probabilities ?

# Classification models predict which category an input belongs to. They often output probabilities to
# show how likely each category is according to the model.
# For example, an image classification model might produce:
# Cat = 0.85
# Dog = 0.10
# Car = 0.05


# This means the model assigns:
# - 85% probability to Cat
# - 10% probability to Dog
# - 5% probability to Car
# The model selects Cat as its prediction because it has the highest probability.
# Probabilities are useful because they allow us to:
# - Compare possible classes.
# - Select the most likely class.
# - Set thresholds for making predictions.
# - Identify predictions that may need human review.
```

```python
# What does a model confidence score represent  ?

# A confidence score indicates how strongly a model favors a particular prediction.
# For example:
# Prediction: Cat
# Confidence score: 0.95

# The model assigns a score of 0.95, or 95%, to the cat prediction.
# However, this score represents real-world likelihood accurately only if the model's
#  probabilities are well calibrated
```

```python
# Is a probability of 0.95 always equal to 95% real-world certainty  ?

# No.A probability of 0.95 means the model assigns a 95% probability to an outcome, but it 
# does not guarantee that the outcome has a 95% real-world chance of being correct.
# For example, if a well-calibrated model makes 1,000 predictions with confidence around 95%, 
# approximately 950 should be correct.
# An overconfident model might make predictions with 95% confidence but achieve a much lower accuracy.
```

```python
# Why can an AI model be overconfident   ?

# An AI model is overconfident when it assigns probabilities that are higher than its actual reliability justifies.
# Common reasons include:
# - Insufficient training data: The model has not learned enough examples.
# - Overfitting: The model learns the training data too closely and struggles with new examples.
# - Unfamiliar inputs: The model encounters data that differs from its training data.
# - Model limitations: Its assumptions or architecture produce poorly calibrated probabilities.
# Example
# Suppose an image classifier sees an unfamiliar animal.
# Cat: 0.99
# Dog: 0.01


# The model strongly favors cat, but the image might actually show a fox.
# The model's high confidence does not guarantee that its prediction is correct.
# One way to address this problem is probability calibration, which adjusts predicted probabilities 
# using validation data to make them better reflect observed outcomes.

# Key takeaway: A model can be confident and still be wrong. Always distinguish the model's predicted 
# probability from real-world certainty.
```
