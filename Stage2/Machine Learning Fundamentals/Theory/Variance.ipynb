```python
# What does variance measure   ?
# Variance measures how spread out the numbers are from their mean (average).
# - Low variance means the values are close to the mean.
# - High variance means the values are more spread out.
# Variance is calculated by finding the average of the squared differences between each value and the mean.
```

```python
# Calculate the variance of 2, 4, 6
# Given:
# X = [2,4,6]

# Step 1: Calculate the mean.
# = 2+4+6/3=4

# Step 2: Subtract the mean from each value and square the differences.
# Value	Difference from mean	Squared difference
# 2	2 − 4 = −2	4
# 4	4 − 4 = 0	0
# 6	6 − 4 = 2	4
# Step 3: Calculate the average of the squared differences.

# Variance=4+0+4/3
# =8/3 approx 2.67

# This is the population variance, where we divide by the total number of values, (N=3).
# If these values are a sample and you want the sample variance, divide by (n-1=2), giving (4).
# Python code
# import numpy as npnumbers = [2, 4, 6]variance = np.var(numbers)print(variance)

# Output:
# 2.6666666666666665


# NumPy's np.var() calculates population variance by default
```

```python
# What does high variance mean   ?
# High variance means the data points are widely spread out from their mean.
# For example:
# A = [1, 5, 9]
# B = [4, 5, 6]
# Both have a mean of 5, but A is more spread out.
# - Variance of A: (10.67)
# - Variance of B: (0.67)
# Therefore, A has higher variance.
```

```python
# What does low variance mean?
# Low variance means the values are close to their mean.
# For example:
# A = [4, 5, 6]
# The values are close to the mean of 5, so the variance is low.
# If every value is identical, the variance is zero.
# identical : means exactly the same value
# For example:
#  A=[5,5,5,5] 

# Every value is identical because every value is 5.

# The mean is:
# 5+5+5+5/4=5 

# Each value is exactly equal to the mean,so there is no difference/deviation from the mean.
# Therefore:

# Variance=0

# Compare this with:

# [4,5,6] 

# These values are not identical, but they are close to each other, so the variance is low, not zero.

# In simple words:

# Identical → exactly the same → [5,5,5]
# Close → slightly different → [4,5,6]
# Very different → [1,5,20] → higher variance
```

```python
# Why is variance important in machine learning   ?
# Variance is important in Machine Learning because it tells us how much the data 
# values vary or spread out from their mean.

# Variance helps us understand the data
# Suppose we have:
# A=[4,5,6] 
# The values are close together → low variance.
# But:
# B=[1,5,20] 
# The values are more spread out → high variance.
# So variance gives us an idea of how much the data changes.


# Variance helps detect important features
# Suppose a dataset has these two features:
# Age:
# [20,21,20,22,21]
# Very little variation → low variance.
# Income:
# [20k,50k,100k,200k,500k]
# Large variation → high variance.
# A feature with zero variance has the same value for every example:
# [5,5,5,5,5]
# Such a feature generally provides no information for distinguishing observations,
#  so it may be removed during preprocessing.

# Variance is important for the bias-variance problem
# In ML, we also talk about variance of a model.
# A model with high variance can learn the training data too closely:
# Training data-->very good
# but may perform poorly on new data.
# This is called overfitting.
# A model with low variance is generally less sensitive to changes in the training data.
# Simple idea
# Variance tells us how much something changes
# For data, it tells us how spread out the values are.
# For a model, it describes how much its predictions can change when the training data changes
```
