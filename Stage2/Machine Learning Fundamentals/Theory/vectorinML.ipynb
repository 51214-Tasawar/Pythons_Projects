```python
# What is a vector in machine learning?
# A vector in machine learning is an ordered list of numbers that represents real-world data like words, 
# images, or house features so an algorithm can process it
# For example:
# V = [25, 175, 70]
# This vector could represent a person's age, height, and weight.

# Why are vectors important for AI? 
# AI and machine learning models use numerical data to learn patterns and make predictions. Vectors 
# provide a way to represent that data as numbers.


# Convert this sentence representation into a vector:
# [0.21, 0.75, 0.13, 0.91]
# What could such a vector represent in an AI system?

# vector is already given 
# This is a 4-dimensional vector, because it contains 4 numerical values.
# In AI, a sentence can be converted into a vector called an embedding.

# For example:
# Sentence:
# "I love learning Python"
#           ↓
#     AI embedding
#           ↓
# [0.21, 0.75, 0.13, 0.91]

# The individual numbers usually do not have a simple human meaning like:
# 0.21 = love
# 0.75 = Python
# Instead, the whole vector represents features/semantic information learned by the AI model.


# What is the difference between a vector and an embedding?

# Vector	                                         Embedding
# An ordered collection of numbers.           	     A vector generated to represent data or meaning.

# Can contain manually selected .	                 Usually generated by a trained model.
# features
# Can represent age, height, and weight.	         Can represent text, images, audio, or other data.

# Does not necessarily represent .	                 Often captures useful patterns or semantic relationships.
# semantic meaning

# The key idea: Every embedding is a vector, but not every vector is an embedding.
```
