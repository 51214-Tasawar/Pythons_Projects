```python
class Parent :
    def __init__(self , name , age):
        self.name = name
        self.age  = age
    def Display(self) :
        print(f"Name : {self.name} Age  : {self.age}")

class child(Parent) :
    pass
e1 = child("Waheed" , 23)
e1.Display()
```

```python
class Animal :
    def __init__(self ,animal , color , price ):
        self.animal = animal
        self.color =  color 
        self.price =  price

class Pet(Animal):
    def __init__(self ,animal,name , color , price):
        super().__init__(animal , color , price)
        self.name = name
    def display(this):
        print(f"Animal is {this.animal}")
        print(f"Name is : {this.name}")
        print(f"Color is : {this.color}")
        print(f"Price is : {this.price}")
p1 = Pet( "Dog","Buddy" , "Black" , 2000)
p1.display()
```

```python
class Human:
    def __init__(self , name , age ):
        self.name = name
        self.age = age
class Student(Human):
    def __init__(self,classsec,name, age , address):
        super().__init__(name, age)
        self.classsec = classsec
        self.address = address
    def display(this):
        print(f"Name is : {this.name}")
        print(f"Age is : {this.age}")
        print(f"Classand Sec : {this.classsec}")
        print(f"Address : {this.address}")

class Emp(Human):
    def __init__(self,name , age , desination , salary):
        super().__init__(name , age)
        self.desination = desination
        self.salary = salary
    def Display(this) :
        print(f"Name is : {this.name}")
        print(f"Age is : {this.age}")
        print(f"Desination is : {this.desination}")
        print(F"Salary is : {this.salary}")

s1 = Student("12th A","Ali" , 22 , "Multan")
s1.display()

e1 = Emp("Ghayur" , 25 , "Junior Dev" , 60000)
e1.Display()
e2 = Emp("Wafadar" , 25 , "Junior Dev" , 60000)
e2.Display()
```

```python
class GrandParent:
    def __init__(self , gname):
        self.gname = gname
class Parent(GrandParent):
    def __init__(self, pname , gname ):
        super().__init__(gname)
        self.pname = pname
class Child(Parent):
    def __init__(self,name , pname , gname):
        super().__init__(pname , gname)
        self.name = name
    def Display(this):
        print(f"Name : {this.name}")
        print(f"Parent name : {this.pname}")
        print(f"Grand Parent : {this.gname}")
c1 = Child("Ali" , "Nasir" , "Riaz")
c1.Display()
```
