```python
user = {
    "name" : "Ali" ,
    "age" : 20 ,
    "Add" : {
        "city" : "Muzaffargarh" ,
         "Provinance" : "Punjab" , 
         "Country" : "Pakistan"
    }
}
user["Class"] = "9th"
print(user["Add"]["city"])
print(user)
```

```python
for i in user.keys() :
    print(i)
```

```python
for j in user.values():
    print(j)
```

```python
for i ,j in user.items():
    if isinstance(j , dict) :
        for x , z in j.items():
            print(f"{x}  : {z}")
    else :
        print(f"{i} :{j}")
```
