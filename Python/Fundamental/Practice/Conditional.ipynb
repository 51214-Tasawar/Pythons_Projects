```python
a = int(input("Enter Number A "))
b = int(input("Enter Number B "))
Ope = int(input("Enter Operation"))
if Ope <= 0 or Ope > 4 :
    print(-1)
    exit()
if Ope== 1 :
    print(a + b)
elif Ope== 2 :
    print(a - b)
elif Ope == 3 :
    print(a * b)
else :
    print(a//b)
```

```python
math = int(input("Enter Math Marks"))
urdu = int(input("Enter Math Marks"))
eng = int(input("Enter Math Marks"))
obt = math + eng + urdu
total = 300

per =  (obt/total) * 100

if per >=90 and per <=100 :
    print("got A+")
elif per >=80 and per < 90 :
    print("Got A")
elif per >= 70 and per <80 :
    print("got  B+")
elif per >= 60 and per <70 :
    print("got  B")
elif per >= 50 and per <60 :
    print("got  C")
else :
    print("Sorry")
```

```python
month = int(input("Enter Month"))

match month :
    case 1|2|3|10|11|12 :
        print("This is Cold session")
    case 4|5|6|7|8|9 :
        print("This is Hot Session")
    case _:
        print("Invalid Input")
```
